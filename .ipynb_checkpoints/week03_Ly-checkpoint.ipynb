## 1. The test suite

Each test is documented with the physical property it checks and why that property catches a specific bug:

| test | physical property | bug it catches |
|---|---|---|
| `test_ground_state_has_zero_point_energy` | $E_0 = \tfrac12\hbar\omega \ne 0$ (uncertainty principle) | 1 |
| `test_energy_levels_are_equally_spaced_by_hbar_omega` | $E_{n+1}-E_n=\hbar\omega$, linear scaling in $\hbar,\omega$ | 1 |
| `test_wavefunctions_are_unit_normalized[n]` | $\int\lvert\psi_n\rvert^2dx = 1$ for each $n=0\ldots4$ | 4 |
| `test_first_five_states_present_with_correct_node_count` | five states; $\psi_n$ has exactly $n$ nodes; parity $(-1)^n$ | 2 (and the legend off-by-one) |
| `test_probability_density_is_non_negative` | $\lvert\psi\rvert^2 \ge 0$ everywhere, while $\psi_{n\ge1}$ goes negative | 5 |
| `test_plot_label_matches_plotted_quantity` | inspects the *figure* `main()` draws: each curve is identified against the analytic $\psi_n$ / $\lvert\psi_n\rvert^2$ (offset and scale allowed), all five states must be present, every legend $n$ must be the curve's true quantum number, and the y-label must name the quantity drawn | 5 (and 2) |
| `test_wavefunction_satisfies_schrodinger_equation[n]` | $\langle\psi_n\lvert H\rvert\psi_n\rangle = n+\tfrac12$ and $\lVert(H-E)\psi_n\rVert^2=0$ — $\psi_n$ is a genuine eigenstate and `energy_levels()` agrees with it | 1, 3, 4 and anything subtler |

Bug 3 (the hallucinated `scipy.special.hermite_poly`) makes every test that evaluates a wavefunction crash, so it is caught thirteen times over. The last test is the strongest: the first six check *properties* an oscillator eigenstate must have, whereas the seventh applies the Hamiltonian $H=-\tfrac12\,d^2/dx^2+\tfrac12x^2$ numerically and demands zero energy variance, which only a true eigenfunction satisfies. It is also the only test that couples `energy_levels()` to `harmonic_state()`.

The identical code lives in `test_quantum_oscillator.py` for command-line use (`QHO_MODULE=<module> pytest -v test_quantum_oscillator.py`). Here the module under test is bound to the name `qho` so the same tests can be pointed at each script in turn.

````python
# --- helpers shared by all tests (identical to the top of test_quantum_oscillator.py) ---
"""Physics-based pytest suite for the quantum harmonic oscillator script (PHYS 7440, Week 03).

Each test checks a physical property of the harmonic oscillator that any correct
implementation must satisfy, independent of how the code is written. Together they
would have caught all five bugs in ``ai_generated_oscillator.py`` before it was run:

    Bug 1  missing zero-point energy          -> test_ground_state_has_zero_point_energy
    Bug 2  ground state dropped (states[1:])   -> test_first_five_states_present_with_correct_node_count
    Bug 3  hallucinated scipy API              -> every test that calls harmonic_state (crashes)
    Bug 4  wrong normalization constant        -> test_wavefunctions_are_unit_normalized
    Bug 5  psi plotted but labelled as density -> test_probability_density_is_non_negative
                                                  + test_plot_label_matches_plotted_quantity
    (all)  is psi_n really the n-th eigenfunction, and does E_n match it?
                                               -> test_wavefunction_satisfies_schrodinger_equation

Units: hbar = m = omega = 1, so the length scale sqrt(hbar / m omega) = 1 and
psi_n(x) = (2^n n! sqrt(pi))^(-1/2) H_n(x) exp(-x^2 / 2)        [Griffiths 3rd ed. Eq. 2.86]
E_n = (n + 1/2) hbar omega                                       [Griffiths 3rd ed. Eq. 2.62]

Select the module under test with the environment variable QHO_MODULE
(default: ai_generated_oscillator). Example:

    QHO_MODULE=ai_generated_oscillator_fixed pytest -v test_quantum_oscillator.py
"""
import importlib
import os

import matplotlib
import numpy as np
import pytest
from scipy.integrate import simpson

matplotlib.use("Agg")  # never open a window during tests
import matplotlib.pyplot as plt  # noqa: E402


X = np.linspace(-8.0, 8.0, 4001)  # wide enough that psi_4 has decayed to ~1e-13 at the edges
N_STATES = 5


def _states():
    """Return an array of shape (5, len(X)) holding psi_0 .. psi_4 evaluated on X."""
    return np.array([qho.harmonic_state(n, X) for n in range(N_STATES)])


def _count_nodes(psi, floor=1e-6):
    """Count interior sign changes of psi, ignoring the tails where psi is numerically zero."""
    core = psi[np.abs(psi) > floor * np.abs(psi).max()]
    return int(np.sum(np.diff(np.sign(core)) != 0))
````

````python
# The test functions are loaded from test_quantum_oscillator.py so that the notebook and the
# command-line runs use one and the same source of truth. Everything after the module-selection
# lines is the test code; it is exec'd against whichever module is bound to the name `qho`.
_src = open("test_quantum_oscillator.py").read()
TEST_SOURCE = _src.split("# ------", 1)[1]
TEST_SOURCE = "# ------" + TEST_SOURCE
print(f"{TEST_SOURCE.count(chr(10) + 'def test_')} test functions loaded from test_quantum_oscillator.py")
print(TEST_SOURCE)
````

````python
def run_suite(module_name, verbose=True):
    """Import `module_name` fresh, bind it to the global name `qho`, and run the suite with pytest."""
    global qho
    sys.modules.pop(module_name, None)
    qho = importlib.import_module(module_name)
    print(f"=== module under test: {module_name}  ({os.path.relpath(qho.__file__)})")
    ipytest.clean()
    exec(TEST_SOURCE, globals())
    return ipytest.run("-v" if verbose else "-q", "-p", "no:cacheprovider")
````

## 4. Does the suite catch each bug *on its own*?

A suite can "fail on the buggy file" purely because of the crash. To show the tests are genuinely physics-driven, each bug is re-injected **one at a time** into the fixed script and the suite is run against that single-bug mutant. Every mutant must fail; a mutant that passes would mean that bug is not actually covered. Seven further mutants probe the suite's limits. Two are *impostors* built to pass every property test: the probabilists' $He_n$ with a compensating normalization, and a Gaussian of the wrong width renormalized numerically for each $n$ — both are unit-normalized, have the right parity and the right number of nodes, and are not eigenfunctions of $H$. Only the Schrödinger-equation test rejects them, which is why it is in the suite. One mutant hard-codes $\omega=1$ inside `energy_levels`; it is caught by the scaling assertion in `test_energy_levels_are_equally_spaced_by_hbar_omega`, not by the eigen-equation test (whose energies are evaluated at $\omega=1$), which shows that assertion earns its place. One uses `eval_hermitenorm`, the interpreter's "did you mean" suggestion. Three are presentation defects that only the figure test can see: $|\psi|^2$ drawn under a $\psi$ label, a legend shifted by one with the data intact, and only four of the five states plotted. The last column shows what that one test alone reports.

````python
import shutil, tempfile

fixed_src = open("ai_generated_oscillator_fixed.py").read()
mutants = {
    "bug1: no zero-point energy":  fixed_src.replace("(np.arange(n_levels) + 0.5)", "np.arange(n_levels)"),
    "bug2: ground state sliced off": fixed_src.replace(
        "    states = np.array([harmonic_state(n, x) for n in range(5)])\n",
        "    states = np.array([harmonic_state(n, x) for n in range(5)])\n    states = states[1:]\n"),
    "bug3: hallucinated hermite_poly": fixed_src.replace("special.eval_hermite(n, x)", "special.hermite_poly(n, x)"),
    "bug4: norm = 1/sqrt(n!)":       fixed_src.replace(
        "norm = 1.0 / np.sqrt(2.0**n * math.factorial(n) * np.sqrt(np.pi))", "norm = 1.0 / np.sqrt(math.factorial(n))"),
    "bug5: psi labelled as density": fixed_src.replace('plt.ylabel(r"$\\psi_n(x)$")', 'plt.ylabel("Probability density")'),
    "trap: eval_hermitenorm (He_n)": fixed_src.replace("special.eval_hermite(n, x)", "special.eval_hermitenorm(n, x)"),
    "He_n + compensating norm":     fixed_src.replace("special.eval_hermite(n, x)", "special.eval_hermitenorm(n, x) * 2**(n/2)")
                                              .replace("np.exp(-x**2 / 2.0)", "np.exp(-x**2 / 4.0) / 2**0.25"),
    "wrong Gaussian width, renorm": fixed_src.replace(
        "    norm = 1.0 / np.sqrt(2.0**n * math.factorial(n) * np.sqrt(np.pi))\n",
        "    _xg = np.linspace(-14.0, 14.0, 28001)  # renormalize numerically so the norm test still passes\n"
        "    norm = 1.0 / np.sqrt(np.sum((special.eval_hermite(n, _xg) * np.exp(-_xg**2 / 4.0))**2) * (_xg[1] - _xg[0]))\n"
    ).replace("np.exp(-x**2 / 2.0) * herm", "np.exp(-x**2 / 4.0) * herm"),
    "omega ignored in energy_levels": fixed_src.replace("hbar * omega * (np.arange(n_levels) + 0.5)", "hbar * (np.arange(n_levels) + 0.5)"),
    "|psi|^2 drawn under a psi label": fixed_src.replace('plt.plot(x, psi, label=f"n={n}")', 'plt.plot(x, psi**2, label=f"n={n}")'),
    "legend shifted (n+1), data intact": fixed_src.replace('plt.plot(x, psi, label=f"n={n}")', 'plt.plot(x, psi, label=f"n={n+1}")'),
    "only four states plotted":        fixed_src.replace("for n, psi in enumerate(states):", "for n, psi in enumerate(states[:4]):"),
}

def pytest_summary(cwd, *extra):
    r = subprocess.run([sys.executable, "-m", "pytest", "-q", "--color=no", "-p", "no:cacheprovider", "test_quantum_oscillator.py", *extra],
                       cwd=cwd, env={**os.environ, "QHO_MODULE": "qho_mutant", "MPLBACKEND": "Agg", "NO_COLOR": "1"},
                       capture_output=True, text=True)
    summary = r.stdout.strip().splitlines()[-1].strip("= ")
    return re.sub(r",? \d+ deselected| in [\d.]+s", "", summary)

print(f"{'mutant':34s} {'full suite':22s} {'Schrödinger test alone':28s} caught?")
print("-" * 96)
for name, src in mutants.items():
    assert src != fixed_src, f"mutation did not apply: {name}"
    with tempfile.TemporaryDirectory() as d:
        open(os.path.join(d, "qho_mutant.py"), "w").write(src)
        shutil.copy("test_quantum_oscillator.py", d)
        full = pytest_summary(d)
        t7 = pytest_summary(d, "-k", "schrodinger")
    print(f"{name:34s} {full:22s} {t7:28s} {'yes' if 'failed' in full else 'NO'}")
````

Here is the test suite:

```python
"""Physics-based pytest suite for the quantum harmonic oscillator script (PHYS 7440, Week 03).

Each test checks a physical property of the harmonic oscillator that any correct
implementation must satisfy, independent of how the code is written. Together they
would have caught all five bugs in ``ai_generated_oscillator.py`` before it was run:

    Bug 1  missing zero-point energy          -> test_ground_state_has_zero_point_energy
    Bug 2  ground state dropped (states[1:])   -> test_first_five_states_present_with_correct_node_count
    Bug 3  hallucinated scipy API              -> every test that calls harmonic_state (crashes)
    Bug 4  wrong normalization constant        -> test_wavefunctions_are_unit_normalized
    Bug 5  psi plotted but labelled as density -> test_probability_density_is_non_negative
                                                  + test_plot_label_matches_plotted_quantity

Units: hbar = m = omega = 1, so the length scale sqrt(hbar / m omega) = 1 and
psi_n(x) = (2^n n! sqrt(pi))^(-1/2) H_n(x) exp(-x^2 / 2)        [Griffiths 3rd ed. Eq. 2.86]
E_n = (n + 1/2) hbar omega                                       [Griffiths 3rd ed. Eq. 2.62]

Select the module under test with the environment variable QHO_MODULE
(default: ai_generated_oscillator). Example:

    QHO_MODULE=ai_generated_oscillator_fixed pytest -v test_quantum_oscillator.py
"""
import importlib
import os

import matplotlib
import numpy as np
import pytest
from scipy.integrate import simpson

matplotlib.use("Agg")  # never open a window during tests
import matplotlib.pyplot as plt  # noqa: E402

MODULE_NAME = os.environ.get("QHO_MODULE", "ai_generated_oscillator")
qho = importlib.import_module(MODULE_NAME)

X = np.linspace(-8.0, 8.0, 4001)  # wide enough that psi_4 has decayed to ~1e-13 at the edges
N_STATES = 5


def _states():
    """Return an array of shape (5, len(X)) holding psi_0 .. psi_4 evaluated on X."""
    return np.array([qho.harmonic_state(n, X) for n in range(N_STATES)])


def _count_nodes(psi, floor=1e-6):
    """Count interior sign changes of psi, ignoring the tails where psi is numerically zero."""
    core = psi[np.abs(psi) > floor * np.abs(psi).max()]
    return int(np.sum(np.diff(np.sign(core)) != 0))


# --------------------------------------------------------------------------------------
# Test 1 — Bug 1
# --------------------------------------------------------------------------------------
def test_ground_state_has_zero_point_energy():
    """Physical property: the ground state energy is E_0 = (1/2) hbar omega, never zero.

    Why: the uncertainty principle forbids a particle from sitting at rest at the bottom
    of the well (x = 0 and p = 0 simultaneously), so the lowest allowed energy is strictly
    positive. Griffiths 3rd ed. Eq. 2.62 gives E_n = (n + 1/2) hbar omega. A formula that
    omits the +1/2 returns E_0 = 0, which is unphysical even though the level *spacing* is
    still correct - exactly the kind of error that looks fine at a glance.
    """
    E = np.asarray(qho.energy_levels(hbar=1.0, omega=1.0, n_levels=N_STATES), dtype=float)
    assert E.shape == (N_STATES,), f"expected {N_STATES} energy levels, got shape {E.shape}"
    assert E[0] == pytest.approx(0.5), f"E_0 should be 0.5 hbar*omega (zero-point energy), got {E[0]}"
    assert np.allclose(E, np.arange(N_STATES) + 0.5), f"E_n should be n + 1/2, got {E}"


# --------------------------------------------------------------------------------------
# Test 2 — Bug 1 (independent check of the spectrum's structure)
# --------------------------------------------------------------------------------------
def test_energy_levels_are_equally_spaced_by_hbar_omega():
    """Physical property: consecutive harmonic-oscillator levels differ by exactly hbar omega,
    and the whole spectrum scales linearly with both hbar and omega.

    Why: the equal spacing is the defining feature of the oscillator spectrum (it is what
    makes the ladder operators work, Griffiths Sec. 2.3.1). Checking the spacing separately
    from E_0 distinguishes 'wrong offset' from 'wrong formula altogether', and checking the
    scaling catches implementations that hard-code hbar = omega = 1.
    """
    E = np.asarray(qho.energy_levels(hbar=1.0, omega=1.0, n_levels=N_STATES), dtype=float)
    assert np.allclose(np.diff(E), 1.0), f"level spacing should be hbar*omega = 1, got {np.diff(E)}"

    E_scaled = np.asarray(qho.energy_levels(hbar=2.0, omega=3.0, n_levels=N_STATES), dtype=float)
    assert np.allclose(E_scaled, 6.0 * (np.arange(N_STATES) + 0.5)), (
        "energy_levels must scale as hbar*omega*(n + 1/2) for general hbar, omega"
    )


# --------------------------------------------------------------------------------------
# Test 3 — Bug 4
# --------------------------------------------------------------------------------------
@pytest.mark.parametrize("n", range(N_STATES))
def test_wavefunctions_are_unit_normalized(n):
    """Physical property: the integral of |psi_n|^2 over all space equals 1 for every n.

    Why: |psi|^2 is a probability density, so the total probability of finding the particle
    *somewhere* must be exactly 1 (Griffiths Eq. 1.20). The correct prefactor is
    (2^n n! sqrt(pi))^(-1/2) (Eq. 2.86 with hbar = m = omega = 1). Dropping the 2^n and
    sqrt(pi) factors gives integrals of 1.77, 3.54, 7.09, 14.2, 28.4 for n = 0..4 - an error
    that grows exponentially with n, so it is tested for each state separately.
    """
    psi = np.asarray(qho.harmonic_state(n, X), dtype=float)
    total_probability = simpson(psi**2, x=X)
    assert total_probability == pytest.approx(1.0, abs=1e-6), (
        f"integral of |psi_{n}|^2 dx = {total_probability:.6f}, expected 1"
    )


# --------------------------------------------------------------------------------------
# Test 4 — Bug 2 (and the legend off-by-one that comes with it)
# --------------------------------------------------------------------------------------
def test_first_five_states_present_with_correct_node_count():
    """Physical property: the set of 'first five states' must contain psi_0 .. psi_4, and the
    n-th eigenfunction has exactly n nodes (interior zero crossings).

    Why: the node theorem for 1-D bound states (Griffiths Sec. 2.3, Fig. 2.7) says the ground
    state has no nodes and each successive level gains one. So a node count is a fingerprint
    of the quantum number that does not depend on labels or array indices. If the ground
    state is sliced away (states[1:]) or the labels are shifted by one, the state in position
    k has k + 1 nodes instead of k and this test fails. It also checks the ground state is
    strictly positive everywhere, which only psi_0 satisfies.
    """
    states = _states()
    assert states.shape[0] == N_STATES, f"expected {N_STATES} states, got {states.shape[0]}"

    nodes = [_count_nodes(psi) for psi in states]
    assert nodes == list(range(N_STATES)), (
        f"state k must have k nodes; got node counts {nodes} for k = 0..{N_STATES - 1}"
    )
    assert np.all(states[0] > 0), "the ground state psi_0 must have no sign change"

    # parity check as a second, independent fingerprint: psi_n(-x) = (-1)^n psi_n(x)
    for n, psi in enumerate(states):
        assert np.allclose(psi[::-1], (-1) ** n * psi, atol=1e-12), f"psi_{n} has the wrong parity"


# --------------------------------------------------------------------------------------
# Test 5 — Bug 5
# --------------------------------------------------------------------------------------
def test_probability_density_is_non_negative():
    """Physical property: a probability density can never be negative, so whatever is
    presented as 'probability density' must satisfy |psi_n|^2 >= 0 at every x.

    Why: psi_n itself is real and changes sign n times, dipping to about -0.64 for n = 1.
    Plotting psi under a 'Probability density' axis label therefore shows negative
    probabilities. This test computes the density explicitly and also confirms that the raw
    wavefunction for n >= 1 does go negative - i.e. that psi and |psi|^2 are genuinely
    different quantities, so the plot must commit to one of them.
    """
    states = _states()
    density = states**2
    assert np.all(density >= 0.0), "|psi|^2 must be non-negative everywhere"
    assert np.all(density.max(axis=1) > 0.0), "each density must be non-trivial"

    for n in range(1, N_STATES):
        assert states[n].min() < 0.0, (
            f"psi_{n} should take negative values; if it does not, it is not a wavefunction"
        )


# --------------------------------------------------------------------------------------
# Test 6 — Bug 5 (checks the figure the script actually produces)
# --------------------------------------------------------------------------------------
def test_plot_label_matches_plotted_quantity(monkeypatch):
    """Physical property: the y-axis label must describe the quantity that is plotted.
    If the label says 'probability density', every curve must be >= 0; if the curves go
    negative, the label must refer to the wavefunction psi.

    Why: this is the output-quantity bug in its natural habitat - the figure. Running
    main() with plt.show() suppressed and inspecting the live Axes catches the mismatch
    regardless of which way the author chooses to resolve it. It also verifies that all
    five curves are present and that the legend labels agree with the node counts, so the
    n=0..4 labelling is checked on the plot itself rather than in the data array.
    """
    if not hasattr(qho, "main"):
        pytest.skip("module has no main(); nothing is plotted")

    captured = {}
    monkeypatch.setattr(plt, "show", lambda *a, **k: captured.setdefault("fig", plt.gcf()))
    plt.close("all")
    qho.main()
    fig = captured.get("fig", plt.gcf())
    ax = fig.axes[0]
    lines = [ln for ln in ax.get_lines() if len(ln.get_xdata()) > 10]
    assert len(lines) == N_STATES, f"expected {N_STATES} plotted states, found {len(lines)}"

    ylabel = ax.get_ylabel()
    says_density = "density" in ylabel.lower() or "probab" in ylabel.lower()
    goes_negative = any(np.min(ln.get_ydata()) < -1e-9 for ln in lines)
    assert not (says_density and goes_negative), (
        f"y-axis says {ylabel!r} but plotted curves go negative: that is psi, not |psi|^2"
    )

    # legend label n must match the node count of the curve it is attached to
    for ln in lines:
        label = ln.get_label()
        digits = "".join(ch for ch in label if ch.isdigit())
        assert digits, f"legend label {label!r} does not state the quantum number"
        n_label = int(digits[0])
        y = np.asarray(ln.get_ydata(), dtype=float)
        expected_nodes = n_label  # true whether psi or |psi|^2 is plotted, since |psi|^2 touches zero at the same points
        measured = _count_nodes(y) if goes_negative else int(
            np.sum((y[1:-1] < y[:-2]) & (y[1:-1] < y[2:]) & (y[1:-1] < 1e-6 * y.max()))
        )
        assert measured == expected_nodes, (
            f"curve labelled {label!r} has {measured} nodes, so it is really n={measured}"
        )
    plt.close(fig)
```

*Note on provenance:* the prompt above embeds the suite as it stood when it was sent (tests 1–6, ten cases). The seventh test — the Schrödinger-equation check — was written afterwards, specifically to close the gap identified in §5.4 below. `ai_regenerated.py` is Claude's output **unchanged**; it passes the enlarged suite without any edit, which is a stronger statement than passing the tests it was shown.

````python
%%writefile ai_regenerated.py
"""One-dimensional quantum harmonic oscillator: energy levels and eigenfunctions.

Formulas (Griffiths, Introduction to Quantum Mechanics, 3rd ed., ch. 2):
    E_n      = (n + 1/2) hbar omega                                    [Eq. 2.62]
    psi_n(x) = (m omega / (pi hbar))^(1/4) (2^n n!)^(-1/2) H_n(xi) exp(-xi^2 / 2),
               xi = sqrt(m omega / hbar) x                              [Eq. 2.86]
With hbar = m = omega = 1 this reduces to
    psi_n(x) = (2^n n! sqrt(pi))^(-1/2) H_n(x) exp(-x^2 / 2).

H_n are the *physicists'* Hermite polynomials (H_0 = 1, H_1 = 2x, H_2 = 4x^2 - 2, ...),
orthogonal with weight exp(-x^2). These are the ones Griffiths uses (Table 2.1), and they
are what scipy.special.eval_hermite computes. (scipy.special.eval_hermitenorm gives the
probabilists' He_n with weight exp(-x^2/2), which would need a different normalization.)
"""
import math

import matplotlib.pyplot as plt
import numpy as np
from scipy.special import eval_hermite  # physicists' Hermite polynomial H_n(x)


def energy_levels(hbar=1.0, omega=1.0, n_levels=5):
    """Return E_n = (n + 1/2) hbar omega for n = 0 .. n_levels - 1."""
    n = np.arange(n_levels, dtype=float)
    return (n + 0.5) * hbar * omega


def harmonic_state(n, x):
    """Normalized eigenfunction psi_n(x) in units hbar = m = omega = 1."""
    x = np.asarray(x, dtype=float)
    norm = 1.0 / math.sqrt(2.0**n * math.factorial(n) * math.sqrt(math.pi))
    return norm * eval_hermite(n, x) * np.exp(-x**2 / 2.0)


def main():
    n_states = 5
    energies = energy_levels(hbar=1.0, omega=1.0, n_levels=n_states)
    print("First five energy levels (units of hbar*omega):")
    for n, E in enumerate(energies):
        print(f"  E_{n} = {E:.1f}")

    x = np.linspace(-6.0, 6.0, 1201)
    fig, ax = plt.subplots(figsize=(8, 5))
    for n in range(n_states):
        ax.plot(x, harmonic_state(n, x), label=f"n={n}")
    ax.set_xlabel(r"$x$  (units of $\sqrt{\hbar / m\omega}$)")
    ax.set_ylabel(r"Wavefunction $\psi_n(x)$")
    ax.set_title("Quantum harmonic oscillator eigenfunctions")
    ax.legend()
    ax.grid(alpha=0.3)
    fig.tight_layout()
    plt.show()


if __name__ == "__main__":
    main()
````

## 6. Reflection

This exercise reveals that AI-generated physics code fails in a way that is more dangerous than ordinary buggy code: it is *physically literate but unverified*. Claude reproduced the overall architecture of the harmonic-oscillator problem — a Gaussian envelope times a Hermite polynomial, an evenly spaced ladder of energies, a loop that plots five curves — and only one of its five errors (the hallucinated `scipy.special.hermite_poly`) announced itself with a traceback. The other four ran silently and would have produced a figure indistinguishable at a glance from the textbook one, while reporting a ground-state energy of zero, wavefunctions with total probability up to 28, a missing ground state, and "probability densities" that dip negative. Each of those is a violation of a principle a physicist could state in one sentence — the uncertainty principle, conservation of probability, the node theorem, positivity of $|\psi|^2$ — and that is exactly why physics-based tests catch them when interface tests do not: my first test suite (three checks that the functions exist, accept keyword arguments and return arrays) detected only the crash, whereas the seven tests here fail on every bug injected individually, and the Schrödinger-equation test additionally rejects two impostors that satisfy normalization, parity and node count yet are not eigenfunctions. The mutation check in §4 makes this concrete: a suite that fails on the buggy file proves little, because the crash masks everything; a suite that fails on *each single-bug mutant* proves coverage. The broader lesson for AI-assisted scientific computing is that the human's leverage has moved from writing the implementation to specifying the invariants that would falsify it. Written before the code and handed to the model as the prompt, those invariants do double duty: they steer the generation toward correct physics, and they remain as a permanent check on the next regeneration, the next refactor and the next model version. Without them the AI is a source of confidently wrong results that are harder to catch than a crash; with them it becomes a fast, checkable drafting tool.
